# Raw half-life source files

Sources: PEPlife2, PepMSND, DRAMP and Cavaco. Run the download cell if the raw files are missing; then run the four load cells. These cells only acquire and load source files.

```python
# Find the repository and set the existing raw-data directory.
from pathlib import Path
import json
import subprocess
import sys

import pandas as pd
from IPython.display import FileLink, Markdown, display

ROOT = next(
    parent for parent in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (parent / "assets/tools/proteolysis_data_manager").is_dir()
)
DATA_ROOT = ROOT / "data/proteolysis/half_time_datasets"
RAW = DATA_ROOT / "raw"
```

## Download raw files

```python
# Run the library acquisition script for the four requested sources.
# The same DATA_ROOT controls acquisition and the raw files loaded below.
DOWNLOAD_SCRIPT = (
    ROOT
    / "assets/tools/proteolysis_data_manager/scripts/downloads/half_time_data/download_all.py"
)
display(Markdown(f"[Download script]({DOWNLOAD_SCRIPT.relative_to(ROOT)})"))
subprocess.run(
    [
        sys.executable,
        str(DOWNLOAD_SCRIPT),
        "--data-root", str(DATA_ROOT),
        "--source", "peplife2",
        "--source", "pepmsnd",
        "--source", "dramp",
        "--source", "cavaco",
    ],
    cwd=ROOT,
    check=True,
)
```

## Load PEPlife2 raw files

```python
# PEPlife2 publishes two raw API partitions; keep them grouped as one database.
peplife2_raw = {
    "natural": json.loads((RAW / "peplife2/natural.json").read_text()),
    "modified": json.loads((RAW / "peplife2/modified.json").read_text()),
}

# Display the source rows directly; no adapter or normalization is applied.
display(pd.DataFrame(peplife2_raw["natural"]["data"]))
display(pd.DataFrame(peplife2_raw["modified"]["data"]))
```
