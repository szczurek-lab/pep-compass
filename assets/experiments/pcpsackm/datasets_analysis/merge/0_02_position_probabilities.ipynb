# Common positional probability matrices

MEROPS marginal counts and PICS occurrence windows are normalized per enzyme, source stratum and position: p(a,j) = count(a,j)/sum_a count(a,j). For qPISA EV1/EV4, input-substrate windows use positive fractional depletion as a continuous weight under the approximately mixed input assumption. Products and repeated EV6 time points do not add occurrence counts.

The same matrix format is used for every source, with 20 amino-acid columns. These are observed marginal distributions, not absolute probabilities of cleavage. Libraries and overlapping MEROPS populations remain separate; no smoothing or background correction is added.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display

ROOT = next(p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
            if (p / "assets/tools/proteolysis_data_manager/src").is_dir())
sys.path.insert(0, str(ROOT / "assets/tools/proteolysis_data_manager/src"))
from proteolysis_data_manager.integration.merge import ProteolysisDataset
DATABASE = ROOT / "data/proteolysis/merged_datasets/proteolysis.sqlite"
dataset = ProteolysisDataset.read(DATABASE)
probabilities = dataset.position_probabilities()
```

```python
display(probabilities.groupby("source_db", sort=False).agg(
    profiles=("profile_id", "nunique"), position_rows=("position", "size"),
).reset_index())
```

## Inspect one profile

Each row is one P4–P4′ position. Amino-acid columns sum to one when observed weight is positive; missing positions remain missing. Change the source, subset and enzyme code below.

```python
SOURCE = "mmp_pics"
SUBSET = "trypsin"
ENZYME_CODE = "M10.001"
selected = probabilities.loc[
    probabilities.source_db.eq(SOURCE)
    & probabilities.source_dataset.astype(str).eq(SUBSET)
    & probabilities.merops_code.eq(ENZYME_CODE)
].sort_values("position_index")
from proteolysis_data_manager.integration.merge.contract.schema import AA
display(selected[["source_db", "source_dataset", "enzyme_name", "position", "observed_weight", *AA]].round(4))
```
